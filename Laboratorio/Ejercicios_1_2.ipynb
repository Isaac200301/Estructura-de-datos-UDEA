```python
import hashlib
import time

secuencia_entrada = input("Ingresa los 10 números: ")

if len(secuencia_entrada) == 10 and secuencia_entrada.isdigit():
    hash_objetivo = hashlib.sha256(secuencia_entrada.encode()).hexdigest()
    print(f"Hash generado: {hash_objetivo}")
    print("Buscando secuencia original...")

    inicio = time.time()
    encontrado = False

    for i in range(10000000000):
        intento = f"{i:010d}"
        if hashlib.sha256(intento.encode()).hexdigest() == hash_objetivo:
            fin = time.time()
            print(f"Secuencia encontrada: {intento}")
            print(f"Tiempo: {fin - inicio:.2f} segundos")
            encontrado = True
            break

    if not encontrado:
        print("No se encontró la secuencia.")
else:
    print("Error: Deben ser exactamente 10 números.")
```

```python
import hashlib
import random
from itertools import permutations

def hash_tx(data):
    if isinstance(data, str):
        return hashlib.sha256(data.encode()).digest()
    return hashlib.sha256(data).digest()

def merkle_root(transacciones):
    nivel = [hash_tx(tx) for tx in transacciones]

    while len(nivel) > 1:
        siguiente = []
        for i in range(0, len(nivel), 2):
            if i + 1 < len(nivel):
                # Hay pareja: se concatenan y se hashea
                combinado = nivel[i] + nivel[i+1]
                siguiente.append(hashlib.sha256(combinado).digest())
            else:
                # No hay pareja: sube de nivel tal cual (sin duplicar ni hashear de nuevo)
                siguiente.append(nivel[i])
        nivel = siguiente
    return nivel[0].hex()

entrada = input("Ingresa las transacciones separadas por coma: ")
transacciones_originales = [t.strip() for t in entrada.split(",")]

root_objetivo = merkle_root(transacciones_originales)
print(f"\nHash Root generado: {root_objetivo}")

transacciones_desordenadas = transacciones_originales.copy()
random.shuffle(transacciones_desordenadas)
print(f"Transacciones desordenadas: {transacciones_desordenadas}")

print("\nBuscando el orden original...")
encontrado = False
for perm in permutations(transacciones_desordenadas):
    if merkle_root(perm) == root_objetivo:
        print(f"¡Orden encontrado!: {list(perm)}")
        encontrado = True
        break

if not encontrado:
    print("No se pudo determinar el orden.")
```
